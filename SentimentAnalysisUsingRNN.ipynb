```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense, Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
num_words = 10000
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words = num_words)
max_sequence_length = 500
X_train = pad_sequences(X_train, maxlen=max_sequence_length)
X_test = pad_sequences(X_test, maxlen=max_sequence_length)
```

```python
model = Sequential([
    Embedding(input_dim=num_words, output_dim=32),
    SimpleRNN(32),
    Dense(1,activation='sigmoid')
])
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
history = model.fit(X_train, y_train, validation_split=0.2, batch_size=64, epochs=5)
```

```python
test_loss, test_accuracy = model.evaluate(X_test, y_test)
print(f'Test accuracy: {test_accuracy:.2f}')
print(f'Test loss: {test_loss:.2f}')
```

```python
word_index = imdb.get_word_index()

review = input("Enter your movie review: ")
encoded_review = []

for word in review.lower().split():
    index = word_index.get(word,2) + 3

    if index >= 10000:
        index = 2
        
    encoded_review.append(index)

padded_review = pad_sequences([encoded_review], maxlen=500)

prediction = model.predict(padded_review)

if prediction[0][0] >= 0.5:
    print("Positive review")
else:
    print("Negative review")
```
